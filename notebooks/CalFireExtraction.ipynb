```python
import requests
import pandas as pd
```

```python
rows = []
for year in range (2016, 2026):
    url = "https://incidents.fire.ca.gov/umbraco/api/IncidentApi/List"
    params = {
        "inactive" : "true",
        "year" : year
    }

    r = requests.get(url, params=params)
    r.raise_for_status()

    data = r.json()
    for row in data:
        row["api_year"] = year
        rows.append(row)
df = pd.DataFrame(rows)
print(df.shape, df.columns.tolist())
```

```python
df.to_csv("calfire_data_2016_2025.csv", index=False)
```
